```python
#CASO 9
#Declaracion de variables
PorcentajeDescuento: float
MontoBase: float
ImportePagado: float
Vuelto: float
ImporteCobrado: float

#Datos de entrada 1
PorcentajeDescuento = float(input ('Digite el porcentaje de descuento: '))
MontoBase = float(input ('Digite el monto base: '))
#Proceso 1: Calcular e informar el monto a pagar
ImporteCobrado = MontoBase - (MontoBase * PorcentajeDescuento / 100)
#Datos de salida 1
print(f'El monto a pagar es S/{ImporteCobrado:.2f}')

#Datos de entrada 2: Solicitar el pago después de ver el monto
ImportePagado = float(input ('Digite el monto con el que paga: '))
#Proceso 2: Calcular el vuelto
Vuelto = ImportePagado - ImporteCobrado
#Datos de salida
print(f'El vuelto es S/{Vuelto:.2f}')
```

```python
#CASO 8
#Declaracion de variables
LadoA: float
LadoB: float
LadoC: float
AreaRectangulo: float
AreaTriangulo: float
AreaTotal: float

#Datos de entrada
LadoA = float(input('Digite el valor del lado A (Altura total): '))
LadoB = float(input('Digite el valor del lado B (Base): '))
LadoC = float(input('Digite el valor del lado C (Altura menor): '))

#Proceso
# El rectángulo tiene base B y altura C
AreaRectangulo = LadoB * LadoC

# El triángulo tiene base B y altura (A - C)
AreaTriangulo = (LadoB * (LadoA - LadoC)) / 2

# El área total es la suma de ambas áreas
AreaTotal = AreaRectangulo + AreaTriangulo

#Datos de salida
print(f'El área total del terreno es {AreaTotal:.2f} m²')
```
